# 📊 Data Preprocessing

The primary objective of this notebook is to **preprocess** the raw Eur/Usd data. This step is essential to ensure the dataset is structured performing Exploratory Data Analysis (EDA) and Feature Engineering.

## 1. Data Acquisition

We acquire historical hourly data for Eur/Usd from 2011 to the present. To achieve this, we utilize the custom script `src/eurusd_data_extraction.py`, which joins multiple datasets of **Histdata**.

```python
import pandas as pd

# We load the datasets
df_5mins = pd.read_pickle('../data/EURUSD_5M.pkl')
df_1h = pd.read_pickle('../data/EURUSD_1H.pkl')
df_1d = pd.read_pickle('../data/EURUSD_1D.pkl')

# We take a look at the data
df_1h
```

As we can see above, there are 69103 rows, which makes a big amount of data. 

## 2. Data Loading & Inspection

We inspect dataset's structure and data types.

```python
# Data info
df_1h.info()
# Statistical summary
df_1h[['open', 'high', 'low', 'close', 'volume']].describe()
```

There are no null values, so there is no matter to worry about them.

## 3. Dealing with volume

As for every row the volume is zero, we are deleting the column.

```python
df_5mins.drop(columns=['volume'], inplace=True)
df_1h.drop(columns=['volume'], inplace=True)
df_1d.drop(columns=['volume'], inplace=True)    

df_1h
```

## 4. Saving the Processed Dataset

Finally, we save the cleaned and preprocessed DataFrame to a new CSV file. 

```python
df_5mins.to_pickle('../data/EURUSD_5M_preprocessed.pkl')
df_1h.to_pickle('../data/EURUSD_1H_preprocessed.pkl')
df_1d.to_pickle('../data/EURUSD_1D_preprocessed.pkl')
```
